```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import numpy as np
import pandas as pd
```

```python
# classification report emotion NLP
emo = pd.read_csv("/Users/inecuypers/Documents/DTA thesis/finals/emotion_NLP_300.csv")

emotion_true = emo[['true_anger', 'true_anticipation','true_disgust', 'true_fear', 'true_joy', 'true_love', 'true_optimism',
                     'true_pessimism','true_sadness', 'true_surprise', 'true_trust']]

emotion_pred = emo[['pred_anger', 'pred_anticipation','pred_disgust', 'pred_fear', 'pred_joy', 'pred_love', 'pred_optimism',
                     'pred_pessimism','pred_sadness', 'pred_surprise', 'pred_trust']]
emotion_true_np = np.array(emotion_true)
emotion_pred_np = np.array(emotion_pred)
emotion_true = emotion_true[:100]
emotion_pred = emotion_pred[:100]
emr_sklearn_NLP = accuracy_score(emotion_true, emotion_pred)
print("Exact Match Ratio (sklearn):", emr_sklearn_NLP)
print("=== classification report emotion NLP ===")
print(classification_report(emotion_true, emotion_pred, zero_division=0))
```
